```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv(r"D:\PROJECT DATA ANALYSIS\messed_Reseller.csv")
```

```python
df_Reseller = df
df_Reseller.info()  
df_Reseller.isnull().sum()
df_Reseller.head()
```

```python
text_cols = df_Reseller.select_dtypes(include="object").columns
for c in text_cols:
    df_Reseller[c] = df_Reseller[c].str.strip()
    df_Reseller[c] = df_Reseller[c].replace("", np.nan)
```

```python
for col in ["Business Type", "State-Province", "Country-Region"]:
    df_Reseller[col] = df_Reseller[col].str.strip().str.title()
```

```python
import re

def smart_title(text, uppercase_tokens=frozenset()):
    if pd.isna(text):
        return text
    text = text.strip()
    if text != text.lower():
        return text
    fixed = []
    for w in text.split(" "):
        letters = re.sub(r"[^A-Za-z]", "", w).lower()
        fixed.append(w.upper() if letters in uppercase_tokens else w.capitalize())
    return " ".join(fixed)

df_Reseller["Reseller"] = df_Reseller["Reseller"].apply(smart_title)
df_Reseller["City"] = df_Reseller["City"].apply(smart_title)
```

```python
df_Reseller["ResellerKey"] = pd.to_numeric(df_Reseller["ResellerKey"], errors="coerce")
df_Reseller["ResellerKey"] = np.abs(df_Reseller["ResellerKey"])
df_Reseller = df_Reseller[df_Reseller["ResellerKey"].notna()].copy()
df_Reseller["ResellerKey"] = df_Reseller["ResellerKey"].astype("Int64")
```

```python
df_Reseller = df_Reseller.drop_duplicates()
df_Reseller = df_Reseller.drop_duplicates(subset=["ResellerKey"], keep="first")
df_Reseller = df_Reseller.sort_values("ResellerKey").reset_index(drop=True)
```
